```python
import numpy as np

def predict(x, w, b):
    return w * x + b

def mse_loss(y_pred, y_true):
    return np.mean((y_pred - y_true) ** 2)

# Datos de ejemplo
x = np.array([1, 2, 3, 4, 5])        # Datos de entrada (variable independiente)
y_true = np.array([2, 4, 5, 4, 5])   # Datos de salida reales (variable dependiente)

# Parámetros del modelo (pendiente e intersección)
w = 0.8
b = 1.0

# Predecir salidas
y_pred = predict(x, w, b)

# Calcular la pérdida MSE
loss = mse_loss(y_pred, y_true)

print(f"Valores predichos: {y_pred}")
print(f"Pérdida MSE: {loss:.3f}")

# Valores predichos: [1.8 2.6 3.4 4.2 5.]
# Pérdida MSE: 0.920


import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

# Generar tamaños de motor realistas y emisiones de CO2 correspondientes
# 100 tamaños de motor de 1.5 a 4.5
engine_sizes = np.linspace(1.5, 4.5, 100)
slope = 60
intercept = 50

# Agregar ruido aleatorio
noise = np.random.normal(0, 15, size=engine_sizes.shape)
co2_emissions = slope * engine_sizes + intercept + noise

# Crear un DataFrame
df = pd.DataFrame({'Engine_size': engine_sizes,
                   'CO2_Emissions': co2_emissions})

df.head()


plt.figure(figsize=(7, 5))
plt.scatter(df['Engine_size'], df['CO2_Emissions'],
            color='b', edgecolors='b', linewidths=1.5)
plt.xlabel('Tamaño del motor')
plt.ylabel('Emisiones de CO2')
plt.title('Tamaño del motor vs. Emisiones de CO2')
plt.grid(alpha=0.5)
plt.show()


import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Preparar la gráfica
fig, axs = plt.subplots(2, 2, figsize=(10, 8))
legend_properties = {'weight': 'bold'}

# Definir los diferentes parámetros a y b para las gráficas
params = [
    (10, 15, 'r'),
    (20, 25, 'y'),
    (30, 35, 'g'),
    (65, 15, 'k')
]

# Iterar y crear cada subgráfica
for i, ax in enumerate(axs.flatten()):
    a, b, color = params[i]
    x = np.linspace(1.5, 4.5, 100)   # Definir el rango para Engine_size

    # Diagrama de dispersión
    x_scatter = df['Engine_size']
    y_scatter = df['CO2_Emissions']
    ax.scatter(x_scatter, y_scatter, edgecolors='b', linewidths=4.)

    # Línea de regresión lineal
    y = a * x + b
    ax.plot(x, y, color, label=f'a = {a}, b = {b}')

    # Etiquetas y cuadrícula
    ax.set_xlabel('Tamaño del motor')
    ax.set_ylabel('Emisiones de CO2')
    ax.legend(prop=legend_properties)
    ax.grid(alpha=0.5)
    ax.set_title(f'Gráfico {i + 1}')

# Ajustar el diseño
plt.tight_layout()
plt.show()


import numpy as np

def mae_loss(y_pred, y_true):
    """
    Calcula el Error Absoluto Medio (MAE) entre los valores predichos y verdaderos.

    Parámetros:
    y_pred (numpy.ndarray): Valores predichos.
    y_true (numpy.ndarray): Valores verdaderos.

    Devuelve:
    float: El Error Absoluto Medio.

    Genera:
    ValueError: Si las entradas no son matrices NumPy o tienen formas diferentes.
    """
    # Validación de entrada
    if not isinstance(y_pred, np.ndarray) or not isinstance(y_true, np.ndarray):
        raise ValueError("Las entradas deben ser matrices NumPy.")

    if y_pred.shape != y_true.shape:
        raise ValueError("Las matrices de entrada deben tener la misma forma.")

    if y_true.size == 0:
        raise ValueError("Las matrices de entrada deben no estar vacío.")

    # Calcular el error absoluto medio
    loss = np.mean(np.abs(y_pred - y_true))

    return loss


# Datos de ejemplo
y_true = np.array([3.0, -0.5, 2.0, 7.0])   # Valores reales
y_pred = np.array([2.5, 0.0, 2.0, 8.0])    # Valores predichos

# Calcular MAE
loss = mae_loss(y_pred, y_true)

print(f"Pérdida MAE: {loss}")
# Pérdida MAE: 0.5


y_true = [3, -0.5, 2, 7]
y_pred = [2.5, 0.0, 2, 8]

try:
    loss = mae_loss(y_pred, y_true)
except ValueError as e:
    print(f"Error: {e}")


import numpy as np

def mse_loss(y_pred, y_true):
    """
    Calcula el error cuadrático medio (MSE) entre los valores predichos y verdaderos.

    Parámetros:
    y_pred (numpy.ndarray): Valores predichos.
    y_true (numpy.ndarray): Valores verdaderos.

    Devuelve:
    float: El error cuadrático medio.

    Genera:
    ValueError: Si las entradas no son matrices NumPy o tienen formas diferentes.
    """
    # Validación de entrada
    if not isinstance(y_pred, np.ndarray) or not isinstance(y_true, np.ndarray):
        raise ValueError("Las entradas deben ser matrices NumPy.")

    if y_pred.shape != y_true.shape:
        raise ValueError("Las matrices de entrada deben tener la misma forma.")

    if y_true.size == 0:
        raise ValueError("Las matrices de entrada deben no estar vacío.")

    # Calcular el error cuadrático medio
    loss = np.mean((y_pred - y_true) ** 2)

    return loss


# Datos de ejemplo
y_true = np.array([3.0, -0.5, 2.0, 7.0])   # Valores reales
y_pred = np.array([2.5, 0.0, 2.0, 8.0])    # Valores predichos

# Calcular MSE
loss = mse_loss(y_pred, y_true)
print(f"MSE Loss: {loss}")

# MSE Loss: 0.375


import numpy as np
import matplotlib.pylab as plt

# Graficar MSE
x_mse = np.arange(-1, 1, 0.01)
y_mse = np.square(x_mse)
plt.plot(x_mse, y_mse, "blue", label="MSE")
plt.grid(True, which="major", alpha=0.5)

# Graficar MAE
x_mae = np.arange(-1, 1, 0.01)
y_mae = np.abs(x_mae)
plt.plot(x_mae, y_mae, "red", label="MAE")
plt.grid(True, which="major", alpha=0.5)

# Etiquetas y título
plt.xlabel('$y_{true} - y_{pred}$', size='x-large')
plt.ylabel('Pérdida', size='x-large')
plt.title('Función de pérdida MAE vs MSE')
plt.legend()

# Mostrar el gráfico
plt.show()


from sklearn.metrics import mean_absolute_error, mean_squared_error

# Ejemplo de valores reales y predichos
y_true = [2, 4, 5, 4, 5]
y_pred = [2.5, 3.8, 4.5, 4.2, 5.1]

# Calcular MAE
mae = mean_absolute_error(y_true, y_pred)
print(f"Error absoluto medio (MAE): {mae}")

# Calcular MSE
mse = mean_squared_error(y_true, y_pred)
print(f"Error cuadrático medio (MSE): {mse}")

# Error absoluto medio (MAE): 0.3
# Error cuadrático medio (MSE): 0.118


from sklearn.metrics import r2_score

# Valores reales
actual = [1, 2, 3, 4, 5]

# Valores predichos
predicted = [1.2, 1.8, 3.2, 3.9, 4.7]

# Calcular el coeficiente de determinación (R²)
r2 = r2_score(actual, predicted)

print("R²: ", r2)
# R²: 0.978


import pandas as pd
import numpy as np

df = pd.read_csv('https://raw.githubusercontent.com/justmarkham/scikit-learn-videos/master/data/Advertising.csv')

# elimina la primera columna del DataFrame.
df = df.iloc[:, 1:]

df.head()

df_new = df[['TV', 'Sales']]
df_new.head()

type(df_new)  # pandas.core.frame.DataFrame

dataset = np.array(df_new)

# Extrae los valores del array en la primera columna y asígnalos a la variable 'x'.
x = dataset[:, 0]
y = dataset[:, 1]

from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

print(x_train.shape)  # (160,)
print(y_train.shape)  # (160,)
print(x_test.shape)   # (40,)
print(y_test.shape)   # (40,)


import matplotlib.pyplot as plt
plt.scatter(x_train, y_train);


x_train = x_train.reshape(-1, 1)
y_train = y_train.reshape(-1, 1)

x_train.shape, y_train.shape
# ((160, 1), (160, 1))


from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(x_train, y_train)

model.coef_, model.intercept_
# 0.04652973, 7.11963843


# Calcular los valores predichos
y_pred = model.predict(x_train)

# Calcular la pérdida (error cuadrático medio)
loss = np.mean((y_train - y_pred) ** 2)
print(loss)   # 10.60


plt.scatter(x_train, y_train);

x = np.linspace(0, 300, 100)

b0 = 7.11
b1 = 0.04

y_hat = b0 + b1*x

plt.plot(x, y_hat, 'r', label=f'b0 = {b0}, b1 = {b1}');
plt.legend()
```
